```python
# draw figures for model fit results
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import scipy as sp
import numpy as np
%matplotlib inline
```

```python
# Draw figures. 3 MPI, 6 MpI: all points with linear regression line, different Hemispheres
def DrawFiguresFitting(FileName, SheetName, NameForSave):
    df = pd.read_excel(FileName, sheet_name=SheetName, header=0)
    sns.set_style()
    sns.set_palette(sns.color_palette('deep'))
    # create the folder
    FolderName = './Figures/' + NameForSave + '/'
    if not os.path.exists(FolderName):
        os.makedirs(FolderName)
    
    # PFF 3 MPI Figure (all points with the linear regression line)
    fig = sns.jointplot(data=df, x="Real3", y="Predict3", kind="reg")
    fig.ax_joint.set_xlabel('Log(Pathologies)', fontweight='bold', fontsize=14)
    fig.ax_joint.set_ylabel('Log(Predicted)', fontweight='bold', fontsize=14)
    fig.plot_marginals(sns.histplot, kde=True)
    plt.show()
    SaveFilePdf = FolderName + NameForSave + ' 3 MPI All Points.pdf'
    SaveFileTif = FolderName + NameForSave + ' 3 MPI All Points.tif'
    fig.savefig(SaveFilePdf, format='pdf')
    fig.savefig(SaveFileTif, format='tif', dpi=300, facecolor='white', transparent=False)

    # PFF 3 MPI Figure (show distributions of different hemispheres)
    g = sns.JointGrid(data=df, x="Real3", y="Predict3", hue="Side")
    fig = g.plot_joint(sns.scatterplot, alpha=.8, edgecolor=".2", linewidth=0, s=50)
    fig.ax_joint.set_xlabel('Log(Pathologies)', fontweight='bold', fontsize=14)
    fig.ax_joint.set_ylabel('Log(Predicted)', fontweight='bold', fontsize=14)
    fig.plot_marginals(sns.histplot, kde=True)
    plt.show()
    SaveFilePdf = FolderName + NameForSave + ' 3 MPI Different hemisphere.pdf'
    SaveFileTif = FolderName + NameForSave + ' 3 MPI Different hemisphere.tif'
    fig.savefig(SaveFilePdf, format='pdf')
    fig.savefig(SaveFileTif, format='tif', dpi=300, facecolor='white', transparent=False)

    # PFF 6 MPI Figure (all points with the linear regression line)
    fig = sns.jointplot(data=df, x="Real6", y="Predict6", kind="reg")
    fig.ax_joint.set_xlabel('Log(Pathologies)', fontweight='bold', fontsize=14)
    fig.ax_joint.set_ylabel('Log(Predicted)', fontweight='bold', fontsize=14)
    fig.plot_marginals(sns.histplot, kde=True)
    plt.show()
    SaveFilePdf = FolderName + NameForSave + ' 6 MPI All Points.pdf'
    SaveFileTif = FolderName + NameForSave + ' 6 MPI All Points.tif'
    fig.savefig(SaveFilePdf, format='pdf')
    fig.savefig(SaveFileTif, format='tif', dpi=300, facecolor='white', transparent=False)


    # PFF 6 MPI Figure (show distributions of different hemispheres)
    g = sns.JointGrid(data=df, x="Real6", y="Predict6", hue="Side")
    fig = g.plot_joint(sns.scatterplot, alpha=.8, edgecolor=".2", linewidth=0, s=50)    
    fig.ax_joint.set_xlabel('Log(Pathologies)', fontweight='bold', fontsize=14)
    fig.ax_joint.set_ylabel('Log(Predicted)', fontweight='bold', fontsize=14)
    fig.plot_marginals(sns.histplot, kde=True)
    plt.show()
    SaveFilePdf = FolderName + NameForSave + ' 6 MPI Different hemisphere.pdf'
    SaveFileTif = FolderName + NameForSave + ' 6 MPI Different hemisphere.tif'
    fig.savefig(SaveFilePdf, format='pdf')
    fig.savefig(SaveFileTif, format='tif', dpi=300, facecolor='white', transparent=False)
```

```python
DrawFiguresFitting('./Datasheet/PFFResults.xlsx', 'Global', 'PFF Global') # global spread model result
DrawFiguresFitting('./Datasheet/PFFResults.xlsx', 'Outgoing', 'PFF Outgoing') # best outgoing gene (Dok5)
DrawFiguresFitting('./Datasheet/PFFResults.xlsx', 'Incoming', 'PFF Incoming') # best incoming gene (Grid2ip)
DrawFiguresFitting('./Datasheet/PFFResults.xlsx', 'Combined', 'PFF Combined') # best combined gene (Pde1a)
```

```python
# Draw figures. LinRLog (concordance correlation coefficient) vs Directionality
def DrawFiguresLinRDir(FileName, SheetName, NameForSave, DirName, LinRLogName):
    df = pd.read_excel(FileName, sheet_name=SheetName, header=0)
    # create the folder
    FolderName = './Figures/LogLinR vs Directionality/'
    if not os.path.exists(FolderName):
        os.makedirs(FolderName)
    
    # Draw figures
    g = sns.JointGrid(data=df, x=DirName, y=LinRLogName,hue="Type",
                      palette=sns.set_palette(sns.color_palette('Dark2')), xlim=(0,1))
    fig = g.plot(sns.scatterplot, sns.histplot, alpha=.2, edgecolor=".2", linewidth=0)
    fig.plot_marginals(sns.histplot, kde=True)
    fig.ax_joint.set_xlabel('Directionality', fontweight='bold', fontsize=14)
    fig.ax_joint.set_ylabel('LinR Log Result', fontweight='bold', fontsize=14)
    plt.show()
    SaveFilePdf = FolderName + NameForSave + ' LogLinR vs Directionality.pdf'
    SaveFileTif = FolderName + NameForSave + ' LogLinR vs Directionality.tif'    
    fig.savefig(SaveFilePdf, format='pdf')
    fig.savefig(SaveFileTif, format='tif', dpi=300, facecolor='white', transparent=False)

```

```python
DrawFiguresLinRDir('./Datasheet/PFFResults.xlsx', 'Directionality', 'Outgoing',
                   'Outgoing Directionality', 'Outgoing LogLinR')
DrawFiguresLinRDir('./Datasheet/PFFResults.xlsx', 'Directionality', 'Incoming',
                   'Incoming Directionality', 'Incoming LogLinR')
DrawFiguresLinRDir('./Datasheet/PFFResults.xlsx', 'Directionality', 'Combined',
                   'Combined Directionality', 'Combined LogLinR')
```
