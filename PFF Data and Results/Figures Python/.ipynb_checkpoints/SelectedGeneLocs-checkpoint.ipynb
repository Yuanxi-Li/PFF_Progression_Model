```python
# draw the distributions of outgoing/incoming/combined, and the locs of snca/gba/vps13
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import scipy as sp
import numpy as np
%matplotlib inline
```

```python
# outgoing distribution
df = pd.read_excel('./Datasheet/PFFResults.xlsx', sheet_name='Directionality', header=0)
sns_histplot = sns.histplot(data=df, x='Outgoing LogLinR', stat='proportion',
                   kde=True, binwidth=0.02,
                   palette=sns.set_palette(sns.color_palette("dark:#5A9_r")))
# snca outgoing result
sns_histplot.axvline(0.656246951, color=[0.706, 0.016, 0.150], linestyle='--') 
# gba outgoing result
sns_histplot.axvline(0.650864375, color=[0.229,0.298,0.754], linestyle='--')
# vps35 outgoing result
sns_histplot.axvline(0.650612227, color=[1, 0.5, 0.2], linestyle='--')
# best outgoing - Dok5
sns_histplot.axvline(0.721860089, color=[0, 0, 0], linestyle='--')

# global model
sns_histplot.axvline(0.622120, color=[0.42, 0.22, 0.02], linestyle='-')
# sns_histplot.set_xlim(0,1)
plt.show()

FolderName = './Figures/Distributions/'
if not os.path.exists(FolderName):
    os.makedirs(FolderName)
    
NameForSave = 'Outgoing'
SaveFileTif = FolderName + NameForSave + ' Distribution.tif'
SaveFilePdf = FolderName + NameForSave + ' Distribution.pdf'
fig = sns_histplot.get_figure()
fig.savefig(SaveFilePdf, format='pdf')
fig.savefig(SaveFileTif, format='tif', dpi=300, facecolor='white', transparent=False)
```
