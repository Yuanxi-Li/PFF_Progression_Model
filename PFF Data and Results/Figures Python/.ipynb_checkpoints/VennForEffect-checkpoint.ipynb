```python
import numpy as np
from matplotlib_venn import venn3, venn3_unweighted
import os
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
def DrawVenn(NameForSave,Labels=['Outgoing', 'Incoming', 'Combined']):
    
    # create the folder
    FolderName = './Figures/Venn For Effect/'
    if not os.path.exists(FolderName):
        os.makedirs(FolderName)
    only_Outgoing = 136
    only_Incoming = 402
    only_Outgoing_Incoming = 0
    Only_Combined = 105
    Only_Outgoing_Combined = 297
    Only_Incoming_Combined = 31
    Outgoing_Incoming_Combined = 67    
    v = venn3_unweighted(subsets=(only_Outgoing, only_Incoming, only_Outgoing_Incoming, Only_Combined,
                   Only_Outgoing_Combined, Only_Incoming_Combined,
                   Outgoing_Incoming_Combined), set_labels=Labels)
    SaveFilePdf = FolderName + NameForSave + '.pdf'
    SaveFileTif = FolderName + NameForSave + '.tif'    
    plt.savefig(SaveFilePdf, format='pdf')
    plt.savefig(SaveFileTif, format='tif', dpi=300, facecolor='white', transparent=False)
```

```python
DrawVenn('Venn For Effect')
```
